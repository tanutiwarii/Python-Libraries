# **Tensors in Pytorch**
A tensor is a container for data that can exist in any number of dimensions.

Think of it as a generalization of:

- **Scalar (0D tensor):** A single number (e.g., 5)
- **Vector (1D tensor):** A list of numbers (e.g., [1, 2, 3])
- **Matrix (2D tensor):** A table of numbers (e.g., a spreadsheet)
- **3D Tensor:** A cube of numbers (e.g., a color image: height × width × RGB channels)
- **4D+ Tensor:** A batch of images, or videos (time × height × width × channels)

```python
import torch
V_data = [1, 2, 3, 4, 5]
V = torch.tensor(V_data)
print(V)
```

#### **Can also create a tensor of random data with a given dimensionality like:**

```python
x = torch.randn((3, 4, 5))
print(x)
```

#### **torch.zeros():** Creates a new Tensor with all elements, initialized as zeros.
#### **torch.ones():**  Creates a new Tensor with all elements, initialized as ones.

```python
y= torch.zeros([3,3])
print(y)

z = torch.ones([3,3])
print(z)
```

#### **torch.full()** and **torch.full_like():**  These functions return a Tensor of the required size filled with required fill_value provided. 

```python
newTensor= torch.full((4, 3), 3.14,dtype= torch.float32)
print(newTensor)

x = torch.full_like(newTensor,3.24, dtype=None )
print(x)
```

### **Tensor attributes:**

Each tensor( torch.Tensor ) has a torch.dtype, torch.device, and torch.layout attributes. 

- **torch.dtype:** A torch.dtype is an object that represents the data type of torch.Tensor. PyTorch has twelve different data types.
- **torch.device:** A torch.device is an object representing the device on which a torch.Tensor is or will be allocated. The torch.device contains a device type ('cpu' or 'cuda') and an optional device ordinal for the device type.

```python
torch.device('cuda:0')
```

```python
x = torch.tensor([1.0, 2.0, 3.0])
print('1D Tensor: \n', x)

y = torch.zeros((3, 3))
print('2D Tensor: \n', y)
```

```python
a = torch.tensor([1.0, 2.0])
b = torch.tensor([3.0, 4.0])

print('Element Wise Addition of a & b: \n', a + b)

print('Matrix Multiplication of a & b: \n',
      torch.matmul(a.view(2, 1), b.view(1, 2)))
```
